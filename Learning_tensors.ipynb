```python
import torch
```

```python
distances = torch.tensor([[3.0],[3.4]])
print(distances.shape) ##Here are 2 samples that is the batch size and 1 feature per sample  
```

# Mixing Type

```python
distances1 = torch.tensor([1.2,1.3,1.4],dtype=torch.float32)
distances2 = torch.tensor([1,2,3],dtype=torch.int64)
mixed_tensor = distances1+distances2
print(f"type:{mixed_tensor.dtype}")
```

```python
#Creating tensors

##1.From List

my_distances=[[1.2,2.3]]
tensor_from_list = torch.tensor(my_distances)
print(tensor_from_list.shape)
##2.From numpy array
import numpy as np
numpy_array = np.array([1,2,2])
tensor_from_numpy=torch.from_numpy(numpy_array)
print(tensor_from_numpy.shape)
```

```python
zeros = torch.zeros(3,3)
print(zeros)
```

```python
ones = torch.ones(3,3)
print(ones)
```

```python
rand = torch.rand(3,3)
print(rand)
```

```python
distance= torch.tensor(23)
print(distance.shape)
```

```python
with_batch=distance.unsqueeze(0)# will add batch dim
print(with_batch)
print(with_batch.shape)
```

```python
ready_for_model=with_batch.unsqueeze(1)
print(ready_for_model)
print(ready_for_model.shape)
```

```python
squeezed=ready_for_model.squeeze()
print(squeezed)
```

```python
## indexing and shaping
predictions = torch.tensor([[1.2],[3.4]],dtype= torch.int16)

#get the first prediction
print(predictions[0])

#get the first 2
print(predictions[:2])

#use .item() to get the py num
print(predictions[0].item())
```
